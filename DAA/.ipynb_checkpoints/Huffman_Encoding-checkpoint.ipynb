```python
import heapq

chars = ['a', 'b', 'c', 'd', 'e', 'f']
freq = [5, 9, 12, 13, 16, 45]

heap = [[freq[i], chars[i]] for i in range(len(chars))]
heapq.heapify(heap)

while len(heap) > 1:
    left = heapq.heappop(heap)
    right = heapq.heappop(heap)

    heapq.heappush(heap, [left[0] + right[0],
                          left[1] + right[1]])

def print_codes(node, code=""):
    if len(node[1]) == 1:
        print(node[1], ":", code)
        return
    print_codes([0, node[1][:len(node[1])//2]], code + "0")
    print_codes([0, node[1][len(node[1])//2:]], code + "1")

print("Huffman Encoding:")
print(heap[0])
```

```python
jobs = [('A', 2, 100), ('B', 1, 19),
        ('C', 2, 27), ('D', 1, 25),
        ('E', 3, 15)]

jobs.sort(key=lambda x: x[2], reverse=True)

max_deadline = max(job[1] for job in jobs)
slots = [None] * max_deadline
profit = 0

for job in jobs:
    for i in range(job[1] - 1, -1, -1):
        if slots[i] is None:
            slots[i] = job[0]
            profit += job[2]
            break

print("Selected Jobs:", [j for j in slots if j])
print("Maximum Profit:", profit)
```
