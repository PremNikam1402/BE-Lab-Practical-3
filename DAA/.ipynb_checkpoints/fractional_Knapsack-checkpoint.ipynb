```python
n = 3
weight = [10, 20, 30]
profit = [60, 100, 120]
capacity = 50

ratio = [profit[i] / weight[i] for i in range(n)]

items = sorted(zip(ratio, weight, profit), reverse=True)

total_profit = 0

for r, w, p in items:
    if capacity >= w:
        total_profit += p
        capacity -= w
    else:
        total_profit += r * capacity
        break

print("Maximum Profit:", total_profit)
```
